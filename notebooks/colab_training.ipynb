```python
# If the repo is not on this machine yet, clone once, then re-run.
# !git clone YOUR_REPO_URL /content/image-captioning

import os
from pathlib import Path


def find_repo_root() -> Path:
    """Works when cwd is repo root or notebooks/."""
    cwd = Path.cwd().resolve()
    for p in [cwd, *cwd.parents]:
        if (p / "train.py").is_file():
            return p
    fallback = Path("/content/image-captioning")
    if (fallback / "train.py").is_file():
        return fallback
    raise FileNotFoundError(
        "Could not find train.py. Clone the repo or open the notebook from the project."
    )


REPO_ROOT = find_repo_root()
os.chdir(REPO_ROOT)
print("REPO_ROOT:", REPO_ROOT)
```
